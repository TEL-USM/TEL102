# Introducción a Linux

* Linux es un sistema operativo con grandes ventajas
    * Seguridad
    * Control y acceso abierto
    * Bajo costo
    * Buen rendimiento
    * Compatibilidad con la red

* La consola linux permite acceder a un conjunto de herramientas de línea de comando que puede acelerar considerablemente la interacción con el sistema.

```python
ls
```

* **rm**: Elimina archivos

```python
rm nuevo_archivo.txt
```

```python
ls
```
