# Introducción a C++

C++ es un lenguaje de programación de propósito general, desarrollado por Bjarne Stroustrup como una extensión del lenguaje C. Combina las características de la programación de bajo nivel con las de alto nivel, permitiendo a los desarrolladores escribir programas eficientes y robustos.

## Características de C++
- **Orientado a objetos:** Soporta conceptos como clases, objetos, herencia, y polimorfismo.
- **Eficiencia:** Permite el control directo sobre el hardware y la gestión de memoria.
- **Flexibilidad:** Se puede utilizar para desarrollar sistemas operativos, juegos, aplicaciones de escritorio, y más.
- **Compatibilidad:** Es compatible con C, permitiendo el uso de bibliotecas y código existente de C.

# Elementos generales

## Tipado Fuerte

Es una característica de los lenguajes de programación donde las reglas de tipos de datos son estrictas

### Ventajas
- **Seguridad**: Ayuda a prevenir errores al asegurar el tipo de dato.
- **Claridad**: Hace que el código sea más fácil de leer y entender.
- **Mantenimiento**: Facilita el mantenimiento del código.

### Desventajas:

- **Complejidad**: Puede requerir más código.
- **Menor Flexibilidad**: Impide que variables cambien su tipo (a diferencia de Python)
- **Curva de Aprendizaje**: Reglas más complejas que el tipado débil.

# Compilación

Traduce a lenguaje de máquina para posterior ejecución (interpretación directa por el procesador)

<center><img src="images/compile1.gif" alt="" width="40%"/></center>

Fuente: http://profesores.elo.utfsm.cl/~agv/elo320/makefile/makefile.html

1. Preprocesamiento
    - Instrucciones del preprocesador (como `#include` y `#define`) se procesan.
    - Genera un archivo con el código fuente modificado.
2. Compilación
    - El código fuente modificado se traduce a código objeto (código máquina).
    - Cada archivo fuente se convierte en un archivo objeto (.o).
3. Enlazado
    - Los archivos objeto se combinan con bibliotecas necesarias para crear un ejecutable.
    - Resuelve referencias entre diferentes archivos objeto.


## ¿Cómo compilar un programa en C++?

```bash
# Preprocesamiento y compilación
g++ -c mi_programa.cpp -o mi_programa.o

# Enlazado  (se pueden enlazar múltiples archivos .o)
g++ mi_programa.o -o mi_programa

# Ejecución
./mi_programa
```
