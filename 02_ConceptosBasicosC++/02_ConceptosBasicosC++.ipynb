# Programación Estructurada

**Paradigma de programación** basa en el **uso extensivo de subrutinas y estructuras de control** con el fin de mejorar la **claridad, calidad y tiempo de desarrollo** del código de programas computacionales.

* En contraste con saltos incodicionales como sentencia **goto** (difícil de seguir mantener)
* Partió con **ALGOL** en los '60

# Elementos generales

* **Datos**
    * **Datos primitivos**: enteros, reales, caracteres, etc.

```c++
int contador = 10;
std::string nombre = "Eduardo";
float PI = 3.141592654;
```

# Elementos generales

* Datos
    * **Estructurados**: agrupar/componer conjuntos de datos del mismo tipo en una unidad (ej. arreglo)

```c++
char semana[5][20] = {"Lu", "Ma", "Mi", "Ju", "Vi"};
int mis_numeros[10] = {1,5,3,4,6,8,3,5,2,10};
```

# Elementos generales

* **Control**
    * **Sentencias**: abstrae conjunto de instrucciones
    * **Estructuras de control**: secuenciación, condición y repetición (ej. if, while).
    * **Abstracción procedural**: procedimientos con un nombre y parámetros (ej. funciones y métodos)
    * **Concurrencias**: computación paralela (ej. procesos y hebras)
* **Tipos de Datos abstractos (TDA)**:
    Agrupa datos y operaciones relacionadas en una unidad (ej. estructuras y clases).
