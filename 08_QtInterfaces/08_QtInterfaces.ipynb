# ¿Qué es Qt?

* Qt (pronunciado çute) es un **framework** de desarrollo de aplicaciones que incluye múltiples bibliotecas
* Fue inicialmente desarrollado para el lenguaje C++, pero en la actualidad es posible utilizarlo en otros lenguajes (Ej: PyQt/PySide para Python)
* Provee mejoras a las bibliotecas nativas de C++ y permite códigos múltiplataforma.
* Algunas plataformas compatibles: Windows, Linux/Unix, MacOS, iOS, Android
* Puede encontrar la última versión de Qt en el siguiente enlace: https://www.qt.io/. 
* La última versión de Qt es Qt6.

# Módulos Qt

Existen dos tipos de módulos en Qt:

* **Módulos Core-Essential**: Son parte de las bibliotecas estándar de Qt. Ejemplo: *QtCore*, *QtGui*, *QtWidgets*, etc.
* **Módulos add-ons (agregados)**: Son módulos externos al desarrollo de Qt (desarrollados por terceros). Ejemplos: *Qt3D*, *QtBluetooth*, *QtSensors*, etc.

# Qt Creator

* Qt, al ser un conjunto de bibliotecas, pueden ser integradas en cualquier IDE que lo permita. Incluso, es posible compilar programas Qt sin IDE.
* Qt Creator es un IDE multiplataforma para desarrollo de programas en C++ creado por el equipo de desarrollo Qt. Está orientado principalmente gráficas fácilmente utilizando las bibliotecas de desarrollo que provee Qt.

# Qt Proyectos

Un proyecto Qt está compuesto por los siguientes elementos:
* **Headers**: Conjunto de archivos con extensión **.h**. Contiene las definiciones de las clases utilizadas en la aplicación.
* **Sources**: Conjunto de archivos con extensión **.cpp**. Contiene las implementaciones de los métodos de clase y funciones utilizadas.
* **Forms**: Conjunto de archivos con extensión **.ui**. Son archivos con formato **XML** que contienen las características y disposición de los distintos *elementos gráficos* de cada pantalla del programa. Estos archivos pueden ser modificados utilizando el *Qt Designer* de forma gráfica o modificando directamente los archivo **.ui**.
* **Archivo .pro**: Es el archivo que contiene las directivas de compilación para el comando **qmake**, comando encargado de generar *makefiles* para desarrollos Qt.

# Qt - HelloWorld: Headers/mainwindow.h

````python
#ifndef MAINWINDOW_H
#define MAINWINDOW_H
#include <QMainWindow>
// Se agrega la clase MainWindow actual al namespace Ui
QT_BEGIN_NAMESPACE
namespace Ui { class MainWindow; }
QT_END_NAMESPACE
// Clase que contiene definiciones de la ventana principal
// Hereda de QMainWindow
class MainWindow : public QMainWindow{
    // Macro que convierte la clase en un QObject
    Q_OBJECT
public:
    // Constructor de ventana principal
    // Una ventana, puede estar asociada a otra ventana/dialog/widget. Si existe
    // esa asociación, *parent != nullptr
    MainWindow(QWidget *parent = nullptr);
    ~MainWindow();
private:
    // Puntero que se hará apuntar a la interfaz gráfica definida en mainwindow.ui
    // Definición de esta clase se encuentra en el archivo "ui_mainwindow.h"
    // (generado automáticamente)
    Ui::MainWindow *ui;
};
#endif // MAINWINDOW_H
````

# Qt - HelloWorld: Sources/main.cpp

````python
#include "mainwindow.h"

#include <QApplication>

int main(int argc, char *argv[])
{
    // Se crea una instancia de aplicación
    QApplication a(argc, argv);
    // Dentro de la aplicación, se crea una instancia de MainWindow
    MainWindow w;
    // Se muestra la ventana principal
    w.show();
    // Se ejecuta la aplicación gráfica
    return a.exec();
}
````

# Qt - HelloWorld: Sources/mainwindow.cpp

````python
#include "mainwindow.h"
#include "ui_mainwindow.h"

// Constructor de ventana principal. Se inicializa con el constructor de
// QMainWindow y su atributo ui con lista de inicialización
MainWindow::MainWindow(QWidget *parent)
    : QMainWindow(parent)
    , ui(new Ui::MainWindow)
{
    // Agrega a MainWindow todos los elementos que se configuran a través del QtDesigner
    ui->setupUi(this);
}

// Destructor de MainWindow. Libera puntero ui una vez se cierra la ventana
MainWindow::~MainWindow()
{
    delete ui;
}
````

# Qt - Qt Charts

* El módulo Qt Charts provee un conjunto de componentes para gráficos fáciles de usar.
* Usa el Qt Graphics View Framework (https://doc.qt.io/qt-6/graphicsview.html) que permite integrar fácilmente gráficos a las interfaces gráficas
* Permite la creación de gráficos de líneas, barras, tortas, polares, áreas, entre otros.
* Debe estar instalado previamente junto con las bibliotecas de Qt para poder utilizarlo.

# Qt - Qt Charts: Configuración

* Incluir en el archivo **.pro** de tu proyecto Qt

```cmake
QT += charts
```

* Incluir en el código la siguiente cabecera

```c++

#include <QtCharts>
```
En versiones anteriores a la versión 6 de Qt, también es necesario agregar lo siguiente
```c++

using namespace QtCharts;
```

* Para grabar imágenes en formato estándar, agregar en su código:

# Qt - Signals y slots

* Permiten la comunicación entre objetos de tipo QObject.
* Una señal (**signal**) es emitida por un objeto para señalar la ocurrencia de un evento o cambio de interés particular.
* Una ranura (**slot**) es un método que es llamado en respuesta  a una señal particular.
* Una clase puede tener tanto signals como slots

# Qt - Signals y slots

* **Signal**
    * Son funciones/métodos de acceso público que **no deben ser implementados y no retornan valores**.
    * Señales son emitidas por un objeto para señalar algún evento de interés (definido por el programador).
* **Slot**
    * Son funciones/métodos que pueden ser llamados normalmente. Su única diferencia es que se pueden conectar con signals.
    * Un slot conectado a una signal es llamado cuando la señal o señales a las que se encuentra conectado son emitidas.

# Qt - Signals y slots - Ejemplo

````python

// Counter.h
#include <QObject>

class Counter: public QObject{
    Q_OBJECT

public:
    Counter(){m_value=0;}
    int value(){return m_value;}

public slots:
    void setValue(int value);

signals:
    void valueChanged(int newValue);

private:
    int m_value;

};

````

# Qt - Signals y slots - Ejemplo

````python

// Counter.cpp
#include "Counter.h"

void Counter::setValue(int value){
    if(value != m_value){
        m_value = value;
        emit valueChanged(value);
    }
}
````

# Qt - Signals y slots - Ejemplo

````python

// main.cpp
#include <QCoreApplication>
#include "Counter.h"
#include <iostream>

int main(int argc, char *argv[])
{
    QCoreApplication a(argc, argv);

    Counter *p = new Counter();
    Counter *q = new Counter();
    QObject::connect(p, SIGNAL(valueChanged(int)), q, SLOT(setValue(int)));

    p->setValue(12);
    std::cout << "El valor de p es " << p->value() << " y el valor de q es " << q->value() << std::endl;

    q->setValue(48);
    std::cout << "El valor de p es " << p->value() << " y el valor de q es " << q->value() << std::endl;

    /* 
        El valor de p es 12 y el valor de q es 12
        El valor de p es 12 y el valor de q es 48
    */
    return a.exec();
}
````

# Qt - Ejemplo: Dial Color

````python

// Headers/rectanglecolor.h

class RectangleColor:public QObject,public QGraphicsRectItem{
    Q_OBJECT
public:
    RectangleColor(int x, int y, int w, int h);

signals:

public slots:
    void changeRed(int);
    void changeGreen(int);
    void changeBlue(int);

private:
    int rojo = 0;
    int verde = 0;
    int azul = 0;
    void updateColor();
};

#endif // RECTANGLECOLOR_H
````

# Qt - Ejemplo: Dial Color

````python

// Sources/rectanglecolor.cpp
#include "rectanglecolor.h"

RectangleColor::RectangleColor(int x, int y, int w, int h) :
    QGraphicsRectItem(x,y,w,h)
{
    updateColor();
}

void RectangleColor::changeRed(int rojo){
    this->rojo = rojo;
    updateColor();
}

void RectangleColor::changeGreen(int verde){
    this->verde = verde;
    updateColor();
}

void RectangleColor::changeBlue(int azul){
    this->azul = azul;
    updateColor();
}

void RectangleColor::updateColor(){
    QColor color(this->rojo, this->verde, this->azul);
    this->setBrush(QBrush(color));
}

````

# Qt - Ejemplo: Dial Color

````python

// Headers/widget.h
#ifndef WIDGET_H
#define WIDGET_H

#include <QWidget>
#include <QGraphicsScene>
#include "rectanglecolor.h"

namespace Ui {
class Widget;
}

class Widget : public QWidget
{
    Q_OBJECT

public:
    explicit Widget(QWidget *parent = 0);
    ~Widget();

private:
    Ui::Widget *ui;
    QGraphicsScene *scene;
    RectangleColor *rect;
};

#endif // WIDGET_H
````

# Qt - Ejemplo: Dial Color

````python

// Sources/widget.cpp
#include "widget.h"
#include "ui_widget.h"

Widget::Widget(QWidget *parent) :
    QWidget(parent),
    ui(new Ui::Widget)
{
    ui->setupUi(this);
    scene = new QGraphicsScene(this);
    ui->graphicsView->setScene(scene);
    rect = new RectangleColor(0, 0, 50, 50);
    scene->addItem(rect);
    connect(ui->dialRojo, SIGNAL(valueChanged(int)), rect, SLOT(changeRed(int)));
    connect(ui->dialVerde, SIGNAL(valueChanged(int)), rect, SLOT(changeGreen(int)));
    connect(ui->dialAzul, SIGNAL(valueChanged(int)), rect, SLOT(changeBlue(int)));
}

Widget::~Widget()
{
    delete ui;
    delete scene;
    delete rect;
}
   
````
