## Estructurando nuestro código
### Caso de estudio:

Realizar un programa que calcule la distancia entre dos puntos ($A$ y $B$) en el plano (coordenadas $(x,y)$), para lo cual se debe usar la formula:

$$
|\overrightarrow{AB}| = \sqrt{(x_{1}-x_{2})^2+(y_{1}-y_{2})^2}
$$

Y no se debe apoyar de ninguna de las funciones de _cmath_.

Para el caso de la raíz cuadrada, usaremos el siguiente aproximación:

$$
\sqrt{x} \approx \frac{x + i^{2}}{2i}
$$

**Donde** $i$ es el número cuyo cuadrado ($i^{2}$) más se acerca a $x$. Para calcular el valor de $i$ debe seguir los siguientes pasos:

1. Sea $i = 1$
2. Se calcula la diferencia ($dif$) como: $dif = |x-i^{2}|$
3. Se debe comparar el valor de $dif$ con el de $|x-((i+1))^{2}|$
4. Si $dif \le |x-((i+1))^{2}|$  se detiene el cálculo, y se obtiene el valor de $i$.
5. En caso contrario, $i=i+1$ y se vuelve al paso 2.

## Makefile

Podemos apoyarnos de un archivo conocido como _Makefile_ el cual permite:
* Crear los archivos necesarios para la creación del ejecutable, y el mismo ejecutable.
* Eliminar todos los archivos utilizados en el camino a la creación del ejecutable.

Por ejemplo, el _makefile_ para el [segundo enfoque](./enfoque2):

```
all: 
	g++ -Wall -c main.cpp mat_simple.cpp header_mat.h header_punto.h
	g++ -Wall -o distancia_exe main.o mat_simple.o

clean:
	rm *.o *.gch distancia_exe
```

Para ejecutarlo se debe realizar lo siguiente:

```
make
```

Para borrar los archivos de apoyo:

```
make clean
```

> Puede tener un nombre diferente, y en dicho caso la instrucción sera ```make -f nombre_archivo``` y ```make clean -f nombre_archivo```

## Namespace

Suponga que debemos realizar el siguiente cambio: ahora nuestro código debe soportar puntos para coordenadas del tipo $(x,y)$, pero también para el caso de $(x,y,z)$.

Un enfoque sería:

* Crear funciones especializadas para caso, y diferenciando sus nombres: ```DistanciaXY``` y ```DistanciaXYZ```.



* Otro enfoque similar es el que se indica en el [tercer enfoque](./enfoque3): uso de un _namespace_ por cada posible escenario (además de mover un poco las funciones).



Lo primero es definir los _namespaces_ necesarios:

```c++
namespace nombre
{
    ...
}
```

Y luego utilizarlos:

* Indicandolo en algún punto del código donde se utiliza:

```c++
using namespace nombre;

```
* O al momento de utilizar alguna definición del mismo:

```c++
nombre::funcion();
```

Por ejemplo:

````c++
namespace ejemplo1{
    void funcion()
    {
        cout << "Namespace 1" <<endl;
    }
}
namespace ejemplo2{
    void funcion()
    {
        cout << "Namespace 2" <<endl;
    }
}
````

Luego:

````c++
using namespace ejemplo2;

funcion();
````

Para usar el otro namespace, se debe indicar (o bien, siempre usar esta forma):

````c++
ejemplo1::funcion();
````

## ```int argc, char** argv```

Muchos de los códigos trabajados incluyen estas definiciones como argumentos de la función _main_:

```c++
int main(int argc, char** argv)
```

Y significan:

* ```int argc```: Indican el número de argumentos **al momento de ejcutarlo** por línea de comando.
* ```char** argv```: Un arreglo de caracteres donde se almacenan los argumentos indicados **al momento de ejecución** por línea de comando.

> El nombre del ejecutable se cuenta en ambos casos

Por ejemplo, considerar la siguiente ejecución:
```bash
 ./main.out 1.2 0.456
```

En este caso:

* argc: 3
* argv:
    * argv[0]: ./main.out
    * argv[1]: 1.2
    * argv[2]: 0.456
